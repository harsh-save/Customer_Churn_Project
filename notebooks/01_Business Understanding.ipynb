## Project Overview
Customer churn refers to customers who discontinue using a company's services.

This project aims to identify the factors influencing customer churn and provide actionable business recommendations using exploratory data analysis and machine learning.

## Business Problem
Customer churn is a major challenge for subscription-based businesses. Acquiring new customers is significantly more expensive than retaining existing ones. Understanding the factors that lead customers to leave can help the buisness improve retention, reduce revenuse loss and enhance customer satisfaction.

## Business Objectives
- Identify key drivers of cutomer churn.
- Analyze customer demographics, services, billing, and contracts.
- Segment customers based on churn.
- Recommed actionable retention strategies.

## Business Questions
- Which customer groups are most likely to churn?
- Does contract type influence churn?
- How does tenure relate to churn?
- Which services reduce churn?
- Do montly charges affect churn?

## Success Criteria

The Project is considered successful if it can:

- Identify major churn drivers
- Discover high-risk customers
- Recommend buisness actions
- Support predictive modeling
- Improve customer rention
